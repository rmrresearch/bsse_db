```python
import re

text = """
    1 O                    8.0000    -0.15218561    -0.00116398     0.00000000
    2 H                    1.0000     0.43776878    -0.76611046     0.00000000
    3 H                    1.0000     0.44759598     0.75607921     0.00000000
"""

pattern = re.compile(
    r"^\s*(\d+)\s+(\w+)\s+([\d\.]+)\s+(-?[\d\.]+)\s+(-?[\d\.]+)\s+(-?[\d\.]+)",
    re.MULTILINE,
)
matches = pattern.findall(text)

for match in matches:
    no, tag, charge, x, y, z = match
    print(f"No: {no}, Tag: {tag}, Charge: {charge}, X: {x}, Y: {y}, Z: {z}")
```
